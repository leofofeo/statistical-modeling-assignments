## 9.4 Probability to odds

##### a. If the probability of an event occurring is 0.8, what are the odds?

The odds are $\frac{p}{1-p} = \frac{.8}{.2} = 4$ (or 4 to 1)

##### b. If the probability of an event occurring is 0.25, what are the odds?

Similarly, the odds are $\frac{p}{1-p} = \frac{.25}{.75} = .333$ (or 1 to 3)

##### c. If the probability of an event occurring is 0.6, what are the odds?

The odds are $\frac{p}{1-p} = \frac{.6}{.4} = 1.5$ (or 3 to 2)

## 9.6 Odds to Probability

##### a. If the odds of an event occurring are 1:3, what is the probability? 
If $P = \frac{\text{successes}}{\text{successes + failures}}$, then the probability of 1:3 odds is $\frac{1}{1+3}= \frac{1}{4} = 0.25$

##### b. If the odds of an event occurring are 5:2, what is the probability? 
Similarly, $P = \frac{5}{7} = .714$

##### c. If the odds of an event occurring are 1:9, what is the probability? 
$P = \frac{1}{10} = .1$, or 10%

##### Consider the fitted logit model from exercise 9.21 How large does a tumor need to be for the estimated probability of metastasis to be at least .8?

We want the odds to reach at least 4:1 (since $\frac{.8}{1.8} = 4$):
$-2.086 + .5117S = ln(4) = 1.3863$

Based on the results, we would say that age is a statistically significant predictor of response, and the drop in deviance (roughly 6.52 on 1df) indicates that adding Age improves fit compared to the intercept only mode.

##### c. Redo parts (a) but using Temp as the single predictor variable
